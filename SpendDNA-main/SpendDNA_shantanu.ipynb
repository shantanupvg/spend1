```python
df["Hour"] = pd.to_datetime(df["Time"], errors="coerce").dt.hour
```

```python
debit_df = df[df["Type"] == "debit"]
```

```python
category_hour = debit_df.pivot_table(
    index="Category",
    columns="Hour",
    values="Amount(Rs.)",
    aggfunc="sum",
    fill_value=0
)
```

```python
category_hour = category_hour.reindex(columns=range(24), fill_value=0)
```

```python

min_value = category_hour.min().min()
max_value = category_hour.max().max()

step = (max_value - min_value) / 5


def intensity(value):
    
    if value == 0:
        return "."
    
    elif value <= min_value + step:
        return "░"
    
    elif value <= min_value + 2 * step:
        return "▒"
    
    elif value <= min_value + 3 * step:
        return "▓"
    
    else:
        return "█"


category_hour_ascii = category_hour.map(intensity)

category_hour_ascii
```

```python
debit_df = df[df["Type"] == "debit"]

category_stats = debit_df.groupby("Category")["Amount(Rs.)"].agg(
    Mean="mean",
    Standard_Deviation="std"
)

# category_stats
```

```python
df["Category_Mean"] = df["Category"].map(category_stats["Mean"])

df["Category_STD"] = df["Category"].map(
    category_stats["Standard_Deviation"]
)


df["Z_Score"] = (
    df["Amount(Rs.)"] - df["Category_Mean"]
) / df["Category_STD"]

df["Anomaly"] = df["Z_Score"] > 2

anomalies = df[df["Anomaly"] == True]
# anomalies
```

```python
category_spending = debit_df.groupby("Category")["Amount(Rs.)"].sum()

category_percentage = (category_spending / total_debits) * 100
```

```python
foodie_percentage = (
    category_percentage.get("Food Delivery", 0)
    + category_percentage.get("Restaurants", 0)
    + category_percentage.get("Cafe", 0)
)

foodie = foodie_percentage > 25
```

```python
quick_commerce_junkie = (
    category_percentage.get("Quick Commerce", 0) > 15
)
```

```python
shopaholic = (
    category_percentage.get("E-commerce", 0) > 15
)
```

```python
investor = (
    category_percentage.get("Investments", 0) > 15
)
```

```python
food_delivery = debit_df[
    debit_df["Category"] == "Food Delivery"
]
late_night_food = food_delivery[
    (food_delivery["Hour"] >= 21) |
    (food_delivery["Hour"] <= 2)
]

late_night_percentage = (
    len(late_night_food) / len(food_delivery)
) * 100

late_night_snacker = late_night_percentage > 50
```

```python
yolo_spender = Savings_rate < 10
```

```python
print("=" * 64)
print("                 SpendDNA REPORT - OMKAR SATPUTE")
print("             6 months - 1,310 transactions - Jan to Jun 2024")
print("=" * 64)

print()
print("EXECUTIVE SUMMARY")
print()

print(f"Total credits   : Rs. {total_credits:,.0f}")
print(f"Total debits    : Rs. {total_debits:,.0f}")
print(f"Net change      : Rs. {Net_savings:,.0f} "
      f"({'overspending' if Net_savings < 0 else 'saving'})")

print(f"Savings rate    : {Savings_rate:.1f}% "
      f"({'BURNING SAVINGS' if Savings_rate < 0 else 'SAVING'})")

print(f"Transactions    : {len(df):,}")
print(f"Unique vendors  : {df[df['Vendor'] != 'Unknown']['Vendor'].nunique()}")



# ==============================================================
# TOP CATEGORIES
# ==============================================================

print()
print("TOP CATEGORIES (% of debit total)")
print()

category_totals = (
    debit_df.groupby("Category")["Amount(Rs.)"].sum().sort_values(ascending=False))

for category, amount in category_totals.head(5).items():

    percentage = (amount / total_debits) * 100

    bar_length = int(percentage / 1.5)
    bar = "#" * bar_length

    print(
        f"{category:<20} "
        f"{bar:<20} "
        f"{percentage:>5.1f}% "
        f"Rs. {amount:,.0f}"
    )



# ==============================================================
# TOP VENDORS
# ==============================================================

print()
print("TOP VENDORS")
print()

vendor_totals = (
    debit_df[debit_df["Vendor"] != "Unknown"].groupby("Vendor")["Amount(Rs.)"].sum().sort_values(ascending=False))

vendor_counts = (
    debit_df[debit_df["Vendor"] != "Unknown"]
    .groupby("Vendor")
    .size()
)

for vendor, amount in vendor_totals.head(5).items():

    count = vendor_counts[vendor]

    print(
        f"{vendor:<25} "
        f"Rs. {amount:,.0f} "
        f"({count:>3} transactions)"
    )



# ==============================================================
# TIME-OF-DAY PATTERNS
# ==============================================================

print()
print("TIME-OF-DAY PATTERNS")
print()

food_delivery = debit_df[
    debit_df["Category"] == "Food Delivery"
]

late_night_food = food_delivery[
    (food_delivery["Hour"] >= 21) |
    (food_delivery["Hour"] <= 2)
]

if len(food_delivery) > 0:
    late_night_percentage = (
        len(late_night_food) / len(food_delivery)
    ) * 100
else:
    late_night_percentage = 0

print(
    f"Food Delivery peaks: "
    f"21:00 - 02:00 "
    f"({late_night_percentage:.1f}% of orders)"
)

cafe = debit_df[
    debit_df["Category"] == "Cafe"
]

if len(cafe) > 0:

    cafe_hour = (
        cafe.groupby("Hour").size().sort_values(ascending=False))

    cafe_peak = cafe_hour.index[0]

    print(
        f"Cafe peak: around {int(cafe_peak):02d}:00"
    )

else:
    print("Cafe peak: No data")


quick_commerce = debit_df[
    debit_df["Category"] == "Quick Commerce"
]

if len(quick_commerce) > 0:

    qc_hour = (
        quick_commerce.groupby("Hour").size().sort_values(ascending=False))

    print(
        f"Quick Commerce peak: "
        f"{int(qc_hour.index[0]):02d}:00"
    )

else:
    print("Quick Commerce: No data")



# ==============================================================
# MONTHLY TREND — FOOD DELIVERY
# ==============================================================

print()
print("MONTHLY TREND (Food Delivery)")
print()

food_monthly = (
    debit_df[debit_df["Category"] == "Food Delivery"].groupby("Month")["Amount(Rs.)"].sum())

month_order = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

food_monthly = food_monthly.reindex(month_order,fill_value=0)

for month, amount in food_monthly.items():

    if food_monthly.max() > 0:
        bar_length = int((amount / food_monthly.max()) * 20)
    else:
        bar_length = 0

    bar = "#" * bar_length

    print(
        f"{month} "
        f"Rs. {amount:,.0f} "
        f"{bar}"
    )



# ==============================================================
# TOP ANOMALIES
# ==============================================================

print()
print("TOP ANOMALIES (z > 2)")
print()

top_anomalies = (
    df[df["Anomaly"] == True].sort_values("Z_Score",ascending=False).head(5))

for _, row in top_anomalies.iterrows():

    date = pd.to_datetime(row["Date"]).strftime("%d %b")

    print(
        f"{date} - "
        f"{row['Vendor']} "
        f"Rs. {row['Amount(Rs.)']:,.0f} "
        f"(z={row['Z_Score']:.2f})"
    )



# ==============================================================
# SPENDING ARCHETYPES
# ==============================================================

print()
print("OMKAR'S SPENDING ARCHETYPES")
print()

archetype_names = {
    "The Foodie": "THE FOODIE",
    "The Quick Commerce Junkie": "THE QUICK COMMERCE JUNKIE",
    "The Shopaholic": "THE SHOPAHOLIC",
    "The Investor": "THE INVESTOR",
    "The Late-Night Snacker": "THE LATE-NIGHT SNACKER",
    "The Cab Commuter": "THE CAB COMMUTER",
    "The Subscription Lover": "THE SUBSCRIPTION LOVER",
    "The YOLO Spender": "THE YOLO SPENDER"
}

for archetype, result in archetypes.items():

    if result:
        print(f" -> {archetype_names[archetype]}")



# ==============================================================
# KEY INSIGHTS
# ==============================================================

print()
print("=" * 64)
print("KEY INSIGHTS")
print("=" * 64)

highest_category = category_totals.index[0]
highest_category_amount = category_totals.iloc[0]

highest_vendor = vendor_totals.index[0]
highest_vendor_amount = vendor_totals.iloc[0]

highest_month = (
    debit_df.groupby("Month")["Amount(Rs.)"]
    .sum()
    .reindex(month_order)
    .idxmax()
)

monthly_total = (
    debit_df.groupby("Month")["Amount(Rs.)"]
    .sum()
    .reindex(month_order)
)

monthly_average = monthly_total.mean()

print()
print(
    f"1. {highest_category} is Omkar's largest spending "
    f"category at Rs. {highest_category_amount:,.0f}."
)

print(
    f"2. {highest_vendor} is the highest-spending recognized "
    f"vendor at Rs. {highest_vendor_amount:,.0f}."
)

print(
    f"3. {highest_month} recorded the highest total spending "
    f"at Rs. {monthly_total[highest_month]:,.0f}."
)

print(
    f"4. Omkar's average monthly spending is approximately "
    f"Rs. {monthly_average:,.0f}."
)

print(
    f"5. The overall savings rate is {Savings_rate:.1f}%, "
    f"indicating {'overspending' if Savings_rate < 0 else 'positive savings'}."
)

print(
    f"6. {len(df[df['Anomaly'] == True])} transactions "
    f"were identified as spending anomalies using z > 2."
)

print()
print("=" * 64)
```
